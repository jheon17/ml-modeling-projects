```python
# 각 컬럼에서 비어 있는 값의 개수를 확인합니다.
df.isna().sum()
```

```python
# 완전히 동일한 중복 행의 개수를 확인합니다.
df.duplicated().sum()
```

```python
# 범주형 컬럼별 고유값 목록과 개수를 확인합니다.
categorical_columns = ["fuel", "seller_type", "transmission", "owner"]

for column in categorical_columns:
    unique_values = df[column].unique()
    print(f"{column} 고유값: {unique_values}")
    print(f"{column} 고유값 개수: {df[column].nunique()}")
    print()
```

```python
# 각 행에서 다섯 컬럼 중 결측치 개수를 계산합니다.
missing_columns = ["mileage", "engine", "max_power", "torque", "seats"]
missing_count_per_row = df[missing_columns].isna().sum(axis=1)
missing_count_per_row.value_counts().sort_index()
```

```python
# 다섯 컬럼이 모두 동시에 결측인 행의 개수를 확인합니다.
df[missing_columns].isna().all(axis=1).sum()
```

```python
# 중복을 제거했을 때 남는 고유한 행의 개수를 확인합니다.
len(df.drop_duplicates())
```

```python
# 13개 전체 컬럼을 기준으로 동일한 행의 등장 횟수를 계산합니다.
row_repeat_counts = df.value_counts(dropna=False)
duplicate_group_counts = row_repeat_counts[row_repeat_counts >= 2]

pd.Series({
    "중복 그룹 개수": len(duplicate_group_counts),
    "최대 반복 횟수": duplicate_group_counts.max(),
})
```

```python
# 중복 그룹이 반복 횟수별로 몇 개인지 확인합니다.
duplicate_group_counts.value_counts().sort_index()
```

```python
# 가장 많이 반복된 중복 그룹 상위 10개를 확인합니다.
top_duplicate_groups = (
    duplicate_group_counts
    .rename("repeat_count")
    .reset_index()
    .head(10)
)

duplicate_display_columns = [
    "name",
    "year",
    "selling_price",
    "km_driven",
    "fuel",
    "seller_type",
    "transmission",
    "owner",
    "repeat_count",
]

top_duplicate_groups[duplicate_display_columns]
```

```python
# 완전 중복인 원본 행 중 앞에서부터 10행을 확인합니다.
duplicate_rows = df[df.duplicated(keep=False)]
duplicate_rows.head(10)
```
